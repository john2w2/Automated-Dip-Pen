```python
import serial
```

```python
box = serial.Serial(port="COM5", baudrate=9600)
```

```python
from time import sleep
```

```python
def enableLight(box):
    box.write(b'\x4f\x7d\x50')
```

```python
def disableLight(box):
    box.write(b'\x4f\x7f\x50')
```

```python
def initBox(box):
    box.write(b'\x57\x02\xff\x50')
    box.write(b'\x57\x03\xfd\x50')
```

```python
# TODO: this is good, move it into the class when we have it
def setIntensity(val, box):
    if (val > 255 or val < 0): raise ValueError("too high!!!!!")
    inv = 255 - val # note, sola inverts intensity, so 0x00 is max while 0xFF is min
    upper = (inv//16) + 240  # need an F to prefix this
    lower = (inv%16) * 16   # need a 0 to follow this, so it's fine
    
    
    box.write(b'\x53\x18\x03\x04%b%b\x50' % (upper.to_bytes(1, 'big'), lower.to_bytes(1, 'big')))
```

```python
def close(box):
    box.close()
```

```python
for i in range(256):
    setIntensity(i, box)
    sleep(0.01)
```
