```python
from OB1 import *
from time import sleep
```

```python
ob = OB1(calibrate=False)
# Time: 3m 31s
```

```python
ob.get_pressure(4)
```

```python
ob.set_pressure(channel=4, pressure= -100)
```

```python
def inThenHold():
    ob.set_pressure(4, -80)
    sleep(1)
    ob.set_pressure(4, 0)
    
```

```python
def dispenseAll():
    ob.set_pressure(4, 80)
    sleep(3)
    ob.set_pressure(4,0)
```

```python
inThenHold()
```

```python
dispenseAll()
```

```python
def inThenOut():
    ob.set_pressure(4, -80)
    sleep(0.5)
    ob.set_pressure(4, 10)
    sleep(2)
    ob.set_pressure(4, 80)
    sleep(3)
    ob.set_pressure(4, 0)
```

```python
inThenOut()
```

```python
def stop():
    ob.set_pressure(4, 0)
```

```python
stop()
```
