```python
import numpy as np
from sklearn.manifold import TSNE
```

```python
import matplotlib.pyplot as plt
```

```python
n = 1000
```

```python
logistic = np.random.logistic(size=n)
normal = np.random.normal(size=n)
uniform = np.random.uniform(size=n)
```

```python
ds = np.concatenate((logistic, normal, uniform))
```

```python
ds.shape
```

```python
tsne = TSNE(n_components=2)
```

```python
x_tsme = tsne.fit_transform(ds.reshape(-1,1))
```

```python
x_tsme.shape
```

```python
plt.figure(figsize=(10,10))
for i in range(3):
    mask = np.squeeze(((np.array([range(3 * n)]) // 1000) == i).transpose())
    plt.scatter(x_tsme[mask, 0], x_tsme[mask, 1])
```
